## Day 1 — Load & trim to the columns we actually need

Column scope comes straight from the 5 research questions (see PROJECT_PLAN.md). This is just trimming, not deep cleaning -- nulls/duplicates/outliers are Day 3.

```python
import pandas as pd
import yaml

with open('../config.yaml') as f:
    config = yaml.safe_load(f)

df = pd.read_csv(config['input_data']['application'])
df.shape
```

### Trim to the 15 columns the research questions need

```python
cols_to_keep = [
    'SK_ID_CURR', 'TARGET', 'CODE_GENDER', 'CNT_CHILDREN',
    'NAME_FAMILY_STATUS', 'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY',
    'NAME_CONTRACT_TYPE', 'NAME_INCOME_TYPE', 'OCCUPATION_TYPE',
    'NAME_EDUCATION_TYPE', 'NAME_HOUSING_TYPE', 'DAYS_BIRTH', 'DAYS_EMPLOYED'
]
df = df[cols_to_keep]
df.shape
```

```python
df.head()
```

### Quick look at what Day 3 cleaning will need to handle

Just checking, not fixing anything yet.

```python
df.isnull().sum()
```

```python
df.dtypes
```

### Sanity check on TARGET

Confirms the ~8% overall default rate used throughout the research questions.

```python
df['TARGET'].value_counts(normalize=True)
```

### Save the trimmed (not yet cleaned) table

```python
df.to_csv(config['output_data']['application'], index=False)
```
