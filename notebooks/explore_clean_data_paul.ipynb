```python
bureau = pd.read_csv(config['input_data']['bureau'])
bureau.shape
```

```python
bureau_columns = [
    'SK_ID_CURR',
    'CREDIT_DAY_OVERDUE',
    'AMT_CREDIT_SUM_OVERDUE'
]

bureau_trimmed_df = bureau[bureau_columns].copy()
bureau_trimmed_df.shape
```

```python
bureau_trimmed_df[
    ['CREDIT_DAY_OVERDUE', 'AMT_CREDIT_SUM_OVERDUE']
].lt(0).sum()
```

```python
bureau_summary_df = (
    bureau_trimmed_df
    .groupby('SK_ID_CURR')
    .agg(
        bureau_record_count=('SK_ID_CURR', 'size'),
        max_days_overdue=('CREDIT_DAY_OVERDUE', 'max'),
        total_overdue_amount=('AMT_CREDIT_SUM_OVERDUE', 'sum')
    )
    .reset_index()
)
```

```python
bureau_summary_df['has_overdue_history'] = (
    (bureau_summary_df['max_days_overdue'] > 0) |
    (bureau_summary_df['total_overdue_amount'] > 0)
).astype(int)
```
